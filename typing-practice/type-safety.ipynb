### TypeVar
Can create a varibale with any type. This is for creating a generic placeholder that persevers the actual types specified when Create the generic type
So when using this we are stating:
- You give a function/method a generic type as sepcified.
- TypeChecker enforces that type.

```python
import tempfile
import subprocess

from IPython.core.magic import register_cell_magic

@register_cell_magic
def typecheck(line, cell):
    with tempfile.NamedTemporaryFile(mode="w", suffix=".py", delete=False) as f:
        f.write(cell)
        f.flush()
        result = subprocess.run(["mypy", f.name], capture_output=True, text=True)
        print(result.stdout)
        print(result.stderr)
```

```python
%%typecheck
from typing import TypeVar, List, reveal_type, Union

# Create a generic type
T = TypeVar("T")


# def sec_item(items: list) -> Union[int | str]:
def sec_item(items: list): # We can use union here but it will give int | str
    return items[1]

def sec_itemT(items: List[T]) -> T:
    return items[1]

nums = [1, 2, 3]
words = ["hi", "hello"]

res1 = sec_item(nums)
resT1 = sec_itemT(nums)

# Hover over the highlighted text to see what the type checker sees
reveal_type(res1)  # Any
reveal_type(resT1) # int

res2 = sec_item(words)
resT2 = sec_itemT(words)

reveal_type(res2) # any
reveal_type(resT2) # str
```

```python
%%typecheck
# Constraining types
from typing import TypeVar, List, reveal_type, Union
T = TypeVar("T", int, float)

def addT(x: T, y: T) -> T:
    return x + y

def add(x: int | float, y: int | float) -> Union[int, float]: # see the inconvience and still the type checkr isnt certain
    return x + y

res = add(2, 3.0)
resT = addT(2, 3.0)
resE = addT("a", "b")
reveal_type(res)
reveal_type(resT)
reveal_type(resE) # Type checker will throw error
```
