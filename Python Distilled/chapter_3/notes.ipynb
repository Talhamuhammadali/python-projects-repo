A common real-world use of `enumerate` is **reporting the position of an item while processing a sequence** — for example, when validating rows in a file and you need to tell the user *which* row failed. ## Exception handling
- Error or break out of normal control flow of a program or script is called an exception
- In python we use `raise <ExceptionType>([value])` to raise an exception with an optional value
- We can catch exceptions using `try` and `except` blocks

```python
# Example of an error

def add(a: int | float, b: int | float):
    if not isinstance(a, (int, float)) or not isinstance(b, (int, float)):
        raise ValueError("Both a and b must be numbers.")
    return a + b

print(add(2, 3))  # Output: 5
print(add(2.5, 3.5))  # Output: 6.

try:
    print(add(2, "three"))  # This will raise a ValueError
except ValueError as e: 
    print(f"Caught an error: {e}")
```

```python
try:
    with open("non_existent_file.txt") as f:
        content = f.read()
except FileNotFoundError as fe:
    print(f"Caught an error: {fe}")
```

In python errors are propgated up the call stack until they are caught. If not caught, pythons defualt handler prints the traceback to standard error `stderr`. The traceback is the chain of how the error propagated up the call stack.


With `try-except` blocks we can avoid programs from exiting, as specified earlier, before the entire control flow is complete. These allow us to catch and handle the errors gracefully by either providing an alternate path for the particular scenario or by logging it as warning and moving on with the rest of the program instead of propagating the error. For scenarios where we want the program to exit, we can re-raise a custom exception based on our code.

```python
def bottom_func():
    return 1 / 0

def middle_func():  
    return bottom_func()

def top_func():
    return middle_func()
try:
    top_func()
except ZeroDivisionError as e:
    print("Caught an error:", e)
    print(e.args)
    print(e.__cause__)
    print(e.__context__)
    print(e.__traceback__)
    
    print("\n--- Traceback ---\n")
    import traceback
    traceback.print_exc()
```

we can add the following clauses to try-except blocks:
- `else` — runs if try block executes succesfully but if exception was raised this is skipped
- `finally` — runs as the final cleanup logic for either cases

```python
def divide(a, b):
    if b == 0:
        raise ZeroDivisionError("Cannot divide by zero.")
    return a / b

def safe_divide_print(a, b):
    answer = None
    try:
        answer = divide(a, b)
    except ZeroDivisionError as e:
        print(f"Caught an error: {e}")
    else:
        print(f"Division successful with answer: {answer}")
        return answer
    finally:
        if answer is not None:
            print("Execution successful.")
        else:
            print("Execution failed.")

safe_divide_print(10, 2)  # Output: Division successful. Execution completed. 5.0
print("\n---\n")
safe_divide_print(10, 0)  # Output: Caught an error
```

### Power of finally
- the finally clause is guaranteed to run by python regradless of the control flow prior to it.
- Best Practice: Never return inside a finally block see the example in the cell below

```python
## Power of finally

def finally_test(finally_return=False):
    try:
        return "The actual return value"
    finally:
        # This block in garunteed final block in the control flow.
        print("This will always execute, regardless of errors.")
        if finally_return:
            return "Returned by finally"

print(finally_test()) # Output: The actual return value, printing the message in finally block as well
print(finally_test(finally_return=True)) # Output: Returned by finally.
```

## Exception Hierarchy
- more than 70 built-in exceptions in Python
- exceptions are orgainized have 'BaseException' as the root of the hierarchy and all other exceptions are derived from it
- `Exception` is the most common base class for user-defined exceptions and most built-in exceptions are derived from `Exception`.


- **Best Practice**: Never catch BaseException unless explicitly required to capture control flow signals.

```python
from pprint import pprint
import builtins

target_builtins = [
    builtin for builtin in dir(builtins) if isinstance(getattr(builtins, builtin), type) and
    
    (issubclass(getattr(builtins, builtin), BaseException) or issubclass(getattr(builtins, builtin), Exception))
]

pprint(target_builtins, indent=2)
print("Total built-in exceptions found:", len(target_builtins))
```

### Custom Exceptions
- Custom exception are derived from the `Exception` class
- This enables the developer to create specific exception for their application without generalizing to the built-in exceptions.
- They can either be a simple str message extension of `Exception` if no body is needed, or we can define an init method to capture additional information about the error.
- if defining init methods it recommended to enrich args attribute with the tuple of all the arguments so that traceback can be more informative when the exception is raised. We  can do this directly using the `super().__init__()` method.
- Naming Convention: Custom exception should be suffixed with `Error` to make it clear that they are exceptions

```python
# Simple example: catching network errors

class NetworkError(Exception):
    """Custom exception for network related errors."""
    pass

def raise_network_error():
    raise NetworkError("Failed to connect to the server.")

try:
    raise_network_error()
except NetworkError as e:
    print(f"Caught an error: {e}")


# We can extend this further for specific cases

class NetworkTimeoutError(NetworkError):
    """Custom exception for network timeout errors."""
    pass

class NetworkHostError(NetworkError):
    """Custom exception for network connection errors."""
    pass

def raise_timeout_error():
    raise NetworkTimeoutError("The connection timed out.")

def raise_host_error():
    raise NetworkHostError("Failed to connect to the host.")

for error_func in [raise_timeout_error, raise_host_error]:
    print(f"\nTesting {error_func.__name__}...")
    try:
        error_func()
    except NetworkError as e:
        if isinstance(e, NetworkTimeoutError):
            print(f"Caught a timeout error: {e}, retrying...")
            continue  # Retry logic could go here
        
        print(f"Caught a network error: {e}")
```

```python
# Problem 2
from dataclasses import dataclass

class InvalidConfigError(Exception):
    """Custom exception for invalid configuration."""
    def __init__(self, message: str, faulty_keys: list):
        super().__init__(message, faulty_keys)
        self.faulty_keys = faulty_keys

@dataclass
class Config:
    host: str
    port: int
    username: str
    password: str
    
    def validate(self):
        faulty_keys = []
        if not self.host:
            faulty_keys.append("host")
        if not (0 < self.port < 65536):
            faulty_keys.append("port")
        if not self.username:
            faulty_keys.append("username")
        if not self.password:
            faulty_keys.append("password")
        
        if faulty_keys:
            raise InvalidConfigError("Invalid configuration detected.", faulty_keys)

# Example usage
try:
    config = Config(host="", port=70000, username="admin", password="")
    config.validate()
except InvalidConfigError as e:
    print(f"Caught an error: {e}")
    print(f"Faulty keys: {e.faulty_keys}")
    print(f"Full error args: {e.args}")
    print("\n--- Traceback ---\n")
    import traceback
    traceback.print_exc()
```

```python
# Problem 3
ACCOUNTS = {
    "blocked_account": {"balance": 200_000, "status": "blocked"},
    "normal_account": {"balance": 20_000, "status": "active"}
}

class BankError(Exception):
    """Base class for bank-related exceptions."""
    pass

class InsufficientFundsError(BankError):
    """ Exception raised when requested amount exceeds current account balance."""
    def __init__(self, message: str, current_balance: float, attempted_withdrawl: float):
        # enrich .args for traceback
        super().__init__(message , current_balance, attempted_withdrawl)
        self.current_balance =  current_balance
        self.attempted_withdrawl = attempted_withdrawl

class AccountLockedError(BankError):
    """Exception raised when user account is locked. """
    pass

class AccountNotFoundError(BankError):
    """ Exception if the account doesnt exist."""
    pass

def withdraw(account: str, withdraw_amount: float):
    """Withdraw the specified amount from the provided account."""
    if account not in ACCOUNTS:
        raise AccountNotFoundError("This account does not exist.")
    account_data = ACCOUNTS[account]
    
    if account_data["status"] == "blocked":
        raise AccountLockedError("Your account is blocked.")
    
    if account_data["balance"] < withdraw_amount:
        raise InsufficientFundsError(
            "Not enough funds for requested withdrawl amount.",
            account_data["balance"],
            withdraw_amount
        )
    
    return f"Withdrew {withdraw_amount} from {account} Successfully."
```

### Chaining Exceptions
We can chain exceptions using ``raise <NewException> from <OriginalException>`` syntax.

```python
class ApplicationError(Exception):
    """Base class for application-specific exceptions."""
    pass

def do_something_risky():
    raise ValueError("Something went wrong in the risky operation.")

def test_chaining_exceptions(disable_chaining=False):
    try:
        do_something_risky()
    except Exception as e:
        # raise 
        e = None if disable_chaining else e
        raise ApplicationError("An error occurred in the application.") from e

try:
    test_chaining_exceptions()
except Exception as e:
    print(f"Caught: {e}")

print("\n--- Special Exceptions ---\n")

try:
    test_chaining_exceptions()
except ApplicationError as e:
    print(f"Caught an ApplicationError: {e}")
    print(f"Original cause (__cause__): {e.__cause__}")
    
print("\n--- Disable Chaining ---\n")

try:
    test_chaining_exceptions(disable_chaining=True)
except ApplicationError as e:
    print(f"Caught an ApplicationError: {e}")
    print(f"Original cause (__cause__): {e.__cause__}")
    print(f"Original context (__context__): {e.__context__}")
```

## Context Managers
- In python when interacting with external resources such as files, databases or other external apis we have a repeating pattern where we establish a connection to the resource, process, clean up connection to the resource. we can use CMs to abstract this using the `with` statement.

- for the with statement to work we should define the `__enter__` and `__exit__` protocols.

- Examples: connecting to a database, opening a file, acquiring a lock, etc.

```python
# counting lines in a file using with statement and context managers

# without using with statement
f = open("../data/access-log.txt")
lc = 0
for line in f:
    lc += 1
f.close()
print(f"Total lines: {lc}")

# with using with statement
lc = 0
with open("../data/access-log.txt") as f:
    for line in f:
        lc += 1
print(f"Total lines: {lc}")
    
```

```python
class SpoiledChild:
    FavoriteFood = ["Chocolate", "ice cream", "noodles"]
    def __init__(self, food):
        self.food = food
        self.bitching = False
        
    def __enter__(self):
        if self.food not in self.FavoriteFood:
            self.bitching = True
        return self
    
    def __exit__(self, exc_type, exc_val, exc_tb):
        pass
    
with SpoiledChild("broccoli") as child:
    if child.bitching:
        print("The child is bitching about the food.")
        
with SpoiledChild("ice cream") as child:
    if child.bitching:
        print("The child is bitching about the food.")
    else:
        print("The child is happy with the food.")
        
```

```python
# Problem
import time

class Timer:
    def __init__(self, label: str, suppress_exceptions: bool = False):
        self.label = label
        self.suppress_exceptions = suppress_exceptions
        
    def __enter__(self):
        self.start = time.time()
        return self

    def __exit__(self, exc_type, exc_val, exc_tb):
        self.end = time.time()
        self.interval = self.end - self.start
        print(f"{self.label}: {self.interval} seconds")
        if self.suppress_exceptions:
            print(exc_type, exc_val, exc_tb)
            return False  # Propagate exceptions
        return True

# Example usage
with Timer("Sleeping"):
    time.sleep(2)

try:
    for flag in [True,  False]:
        with Timer("Risky Operation", suppress_exceptions=flag):
            raise ValueError("An error occurred during the risky operation.")
        print("Risky operation completed successfully.")
    
except Exception as e:
    print(f"Caught an error: {e}")
```

We have two return scenarios for the `__exit__` method:
- returns True: This  means that the exception was either handled or ignored
- return False or None: propagates the exception type value and traceback.

Note: the `__exit__` method protocol always takes the three exeception args; exception type, its value and traceback.
