Using enumerate:
- creates an index and a value for each item in an iterable
- it produces a tuple

## Useful case for `enumerate`

A common real-world use of `enumerate` is **reporting the position of an item while processing a sequence** — for example, when validating rows in a file and you need to tell the user *which* row failed. It keeps the index and the value bound together in a single readable loop, instead of maintaining a manual counter.

---

### Problems to solve

**Problem 1 — Find the positions of a target**
Given a list of words, print the index of every word that equals `"python"` (case-insensitive).

**Problem 2 — Numbered shopping list**
Given a list of items, print a numbered shopping list that starts counting at `1` (not `0`), with each line formatted as `"<n>. <item>"`.


```python
print("--- Problem 1 ---")
words = ["hello", "world", "python", "programming", "PYTHON", "python"]
for idx, w in enumerate(words):
    if w.lower() == "python":
        print(f"{idx}: {w}")

print("--- Problem 2 ---")
shopping_list = ["milk", "eggs", "bread", "butter"]
for idx, item in enumerate(shopping_list, start=1):
    print(f"{idx}. {item}")
```

## Using `zip`:

- when we are iterating over multiple sequences together and processing them we combine create a an iterator using zip
- it produces a tuple with each item from the sequences
- iterator stops on the shortest sequence

### Problems to solve — `zip`

**Problem 1 — Pair students with their scores**
Given two lists, one of student names and one of their test scores, print each line as `"<name>: <score>"`. The lists are guaranteed to be the same length.

**Problem 2 — Mismatched lengths**
Given two lists of *different lengths* (for example, 5 names and 3 scores), use `zip` to print each `name: score` pair. Then in a comment, write down what you observed about the leftover items — does `zip` raise an error, drop them, or do something else? Bonus: try `itertools.zip_longest` and note how it differs.

**Problem 3 — Unzipping**
You are given a single list of `(name, age)` pairs, e.g. `[("alice", 30), ("bob", 25), ("carol", 40)]`. Using `zip` (no manual loops, no list comprehensions), split it into two separate tuples: one containing all the names, one containing all the ages. *(Hint: the `*` unpacking operator is involved.)*


```python
from itertools import zip_longest
print("--- Problem 1 ---")
students = ["Alice", "Bob", "Charlie", "David"]
grades = [85, 92, 78, 90]

for student, grade in zip(students, grades):
    print(f"{student}: {grade}")

print("--- Problem 2 ---")
students = ["Alice", "Bob", "Charlie", "David"]
grades = [85, 92] # two students have missed the test
for student, grade in zip(students, grades):
    print(f"{student}: {grade}") # iterator stops on the shortest sequence
    
for student, grade in zip_longest(students, grades, fillvalue="No Grade"):
    print(f"{student}: {grade}")

print("--- Problem 3 ---")
people = [("Alice", 30), ("Bob", 25), ("Charlie", 35)]
names, ages = zip(*people)
print("Names:", names)
print("Ages:", ages)
```

Something new i learned:
for loops can have an `else` that only executes if loop completes without hitting a `break` statement.

```python
lines = [
    "This is the first line.",
    "This is the second linae.",
    "  ",  # This line is just whitespace
    "This is the third line."
]

for line in lines:
    stripped_line = line.strip()
    print(stripped_line)
    if not stripped_line:
        continue
else:
    # Only executed if the loop iterates exhaustively without hitting a break
    print("All lines processed successfully.")
```

## Exception handling
- Error or break out of normal control flow of a program or script is called an exception
- In python we use `raise <ExceptionType>([value])` to raise an exception with an optional value
- We can catch exceptions using `try` and `except` blocks

```python
# Example of an error

def add(a: int | float, b: int | float):
    if not isinstance(a, (int, float)) or not isinstance(b, (int, float)):
        raise ValueError("Both a and b must be numbers.")
    return a + b

print(add(2, 3))  # Output: 5
print(add(2.5, 3.5))  # Output: 6.

try:
    print(add(2, "three"))  # This will raise a ValueError
except ValueError as e: 
    print(f"Caught an error: {e}")
```

```python
try:
    with open("non_existent_file.txt") as f:
        content = f.read()
except FileNotFoundError as fe:
    print(f"Caught an error: {fe}")
```

In python errors are propgated up the call stack until they are caught. If not caught, pythons defualt handler prints the traceback to standard error `stderr`. The traceback is the chain of how the error propagated up the call stack.


With `try-except` blocks we can avoid programs from exiting, as specified earlier, before the entire control flow is complete. These allow us to catch and handle the errors gracefully by either providing an alternate path for the particular scenario or by logging it as warning and moving on with the rest of the program instead of propagating the error. For scenarios where we want the program to exit, we can re-raise a custom exception based on our code.

```python
def bottom_func():
    return 1 / 0

def middle_func():  
    return bottom_func()

def top_func():
    return middle_func()
try:
    top_func()
except ZeroDivisionError as e:
    print("Caught an error:", e)
    print(e.args)
    print(e.__cause__)
    print(e.__context__)
    print(e.__traceback__)
    
    print("\n--- Traceback ---\n")
    import traceback
    traceback.print_exc()
```

we can add the following clauses to try-except blocks:
- `else` — runs if try block executes succesfully but if exception was raised this is skipped
- `finally` — runs as the final cleanup logic for either cases

```python
def divide(a, b):
    if b == 0:
        raise ZeroDivisionError("Cannot divide by zero.")
    return a / b

def safe_divide_print(a, b):
    answer = None
    try:
        answer = divide(a, b)
    except ZeroDivisionError as e:
        print(f"Caught an error: {e}")
    else:
        print(f"Division successful with answer: {answer}")
        return answer
    finally:
        if answer is not None:
            print("Execution successful.")
        else:
            print("Execution failed.")

safe_divide_print(10, 2)  # Output: Division successful. Execution completed. 5.0
print("\n---\n")
safe_divide_print(10, 0)  # Output: Caught an error
```

### Power of finally
- the finally clause is guaranteed to run by python regradless of the control flow prior to it.
- Best Practice: Never return inside a finally block see the example in the cell below

```python
## Power of finally

def finally_test(finally_return=False):
    try:
        return "The actual return value"
    finally:
        # This block in garunteed final block in the control flow.
        print("This will always execute, regardless of errors.")
        if finally_return:
            return "Returned by finally"

print(finally_test()) # Output: The actual return value, printing the message in finally block as well
print(finally_test(finally_return=True)) # Output: Returned by finally.
```

## Exception Hierarchy
- more than 70 built-in exceptions in Python
- exceptions are orgainized have 'BaseException' as the root of the hierarchy and all other exceptions are derived from it
- `Exception` is the most common base class for user-defined exceptions and most built-in exceptions are derived from `Exception`.


- **Best Practice**: Never catch BaseException unless explicitly required to capture control flow signals.

```python
from pprint import pprint
import builtins

target_builtins = [
    builtin for builtin in dir(builtins) if isinstance(getattr(builtins, builtin), type) and
    
    (issubclass(getattr(builtins, builtin), BaseException) or issubclass(getattr(builtins, builtin), Exception))
]

pprint(target_builtins, indent=2)
print("Total built-in exceptions found:", len(target_builtins))
```

### Introspection
- we can use the `__subclasses__()` to see the exceptions that inherit from a particular exception class
- we can use the `__mro__` attribute to see the method resolution order to walk up the exception hierarchy

```python
print(len(BaseException.__subclasses__()))
print(len(Exception.__subclasses__()))

print(BaseException.__subclasses__())
print(Exception.__subclasses__())
```
