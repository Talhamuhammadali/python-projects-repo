Iterators examples

```python
s = [ (1, 2, 3), (4, 5, 6) ] 
```

```python
for a, b, c in s:
    print(a, b, c)
```

```python
s = [ (1, 2), (3, 4, 5), (6, 7, 8, 9) ] 

print("Various unpacking examples")
for a, *extra, c in s:
    print(a, extra, c)
print("---"*20)
for a, *extra in s:
    print(a, extra)
for first, *rest in s:
    print(first, rest)

# Fails for other values
for a,b in s[:1]:
    print(a, b)
```

```python
# enumeration example

for i,v in enumerate(s):
    print(i, v)
    
    
# we can set the start value for enumeration
for i,v in enumerate(s, start=12):
    print(i, s)
    
    
# zip example
a = [1, 2, 3]
v = ["a", "b", "c"]
for x, y in zip(a, v):
    print(x, y)
```

```python
try:
    with open("not_found.txt") as f:
        for line in f:
            print(line)
except FileNotFoundError as e:
    print("File not found.")
    print(e.args)
    print(e.__cause__)
    print(e.__context__) # We can check it this error was triggered by another error
    print(e.__traceback__)
```
