Compound Investment Calculator

```python
prinicpal = 10000
rate = 0.17
num_years = 10
year = 1

while year <= num_years:
    amount = prinicpal * (1 + rate)
    print(f"Year {year:>1d}: ${amount:.2f}")
    prinicpal = amount
    year += 1
```

```python
# masking 
# Lets say we only care about the first 4 bits of a binary number, we can use a mask to extract those bits

a = 0b10101010
print(f"Integer value: {int(a)}")
mask = 0b11110000
print(f"Mask: {bin(mask)}")
result = a & mask
print(f"Result: {bin(result)}")
result = (a & mask) >> 4
print(f"Result of only the first 4 bits: {bin(result)}")
```

```python
# The walrus operator (:=) 

# Do operations and assign the result to a variable in a single expression
x = 0
while (x := x + 1) <= 5:
    print(x)
```

```python
# Using walrus operator in list comprehensions
numbers = [1, 2, 3, 4, 5]
squared_numbers = [y := x**2 for x in numbers]
print(f"Squared numbers: {squared_numbers}")
```
