Compound Investment Calculator

```python
prinicpal = 10000
rate = 0.17
num_years = 10
year = 1

while year <= num_years:
    amount = prinicpal * (1 + rate)
    print(f"Year {year:>1d}: ${amount:.2f}")
    prinicpal = amount
    year += 1
```

```python
# masking 
# Lets say we only care about the first 4 bits of a binary number, we can use a mask to extract those bits

a = 0b10101010
print(f"Integer value: {int(a)}")
mask = 0b11110000
print(f"Mask: {bin(mask)}")
result = a & mask
print(f"Result: {bin(result)}")
result = (a & mask) >> 4
print(f"Result of only the first 4 bits: {bin(result)}")
```

Object and Classes

- Python is mostly objected based programming language. It has a lot of built-in classes and objects. We can also create our own classes and objects.
- example: int and str are object which we will explore here

```python
a_str = "Hello, World!"
a_int = 42
a_list = [1, 2, 3]

print(dir(a_str))
print(dir(a_int))
print(dir(a_list))

print("==="*5)
common = set(dir(a_str)) & set(dir(a_int)) & set(dir(a_list))
items = list(common)                                                                                                        
print("common items between all: ", end="")                                                                                       
for i in range(0, len(items), 3):    
    print(', '.join(items[i:i+3]))
print("==="*5)
print("Unique to string: ", end="")
unique_str = set(dir(a_str)) - common
for i in range(0, len(unique_str), 3):
    print(', '.join(list(unique_str)[i:i+3]))

print("==="*5)
print("Unique to int: ", end="")
unique_int = set(dir(a_int)) - common
for i in range(0, len(unique_int), 3):
    print(', '.join(list(unique_int)[i:i+3]))
print("==="*5)
print("Unique to list: ", end="")
unique_llst = set(dir(a_list)) - common
for i in range(0, len(unique_llst), 3):
    print(', '.join(list(unique_llst)[i:i+3]))
```
