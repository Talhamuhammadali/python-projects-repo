Python Building Blocks:

- Literal: A value typed directly and not assigned to a variable

```python
# Example of literal integer in Python
print(42)
print(f"binary: {bin(42)}")
print(f"octal: {oct(42)}")
print(f"hexadecimal: {hex(42)}")
print(f"large number with underscores: {1_000_000}")  # Using underscores for readability
# Example of literal float in Python
print("=== Float Literals ===")
print(15000.0)
print(1.5e4)  # Scientific notation for 15000.0
print(0.000015)  # Another way to represent 1.5e-5
print(1.5e-5)  # Scientific notation for 0.000015
print(1.5E-5)  # Scientific notation with uppercase 'E' for 0.000015
# Example of literal string in Python
print("=== String Literals ===")
print("Hello, World!")
# Example of literal boolean in Python
print(True)
print(False)
```

- Expressions: A computation that evaluates to some concrete value. Usually a combination of `Literals`, `Operators`, `names` and `functions/method calls`


```python
# Example of expression covering all
# Area of a circle with radius 5
import math
radius = 5
area = math.pi * radius ** 2  # This expression uses the formula A = πr^2 
print(f"Area of a circle with radius {radius}: {area:.2f}")
print("Variable 'area' is the location where the reference of the expression is stored.")
```

```python
# in place
a = [1, 2, 3, 3]
b = a

a += b  # This modifies 'a' in place by extending it with the elements of 'b'
print(f"After in-place addition, a: {a}, b: {b}")
print(f"Reference level check: a is b: {a is b}")
a = [1, 2, 3, 3]
b = a
a = a + b  # This creates a new list and assigns it to 'a', leaving 'b' unchanged
print(f"After regular addition, a: {a}, b: {b}")
print(f"Reference level check: a is b: {a is b}")
```

Object Comparison
- `==` is equality operator its checks if the contents of object being compared match.
- `is` is and identity operator its checks if the reference in memory is the same
We never compare using is operator unless checking if two objects are the same.

```python
a = [1,3]
b = [1,3]

print("Equality check:", a == b)  # True, because values are equal
print("Identity check:", a is b)  # False, because they are different objects

# for rare cases like integers they both will be same
x = 1
y = 1
print("Equality check for integers:", x == y)  # True, because values are equal
print("Identity check for integers:", x is y)  # True, because small integers are cached by Python and point to the same object
```

```python
set_a = {1, 2, 3}
set_b = {1, 2, 3, 4, 5}
set_c = {11, 12, 13}


def check_subset_superset(set1, set2):
    print(f"Checking if \033[1;36m{set1}\033[0m is a subset of \033[1;36m{set2}\033[0m: {set1 < set2}")
    print(f"Checking if \033[1;36m{set1}\033[0m is a superset of \033[1;36m{set2}\033[0m: {set1 > set2}")
    print(f"Checking if \033[1;36m{set1}\033[0m is a subset or equal to \033[1;3m{set2}\033[0m: {set1 <= set2}")
    print(f"Checking if \033[1;36m{set1}\033[0m is a superset or equal to \033[1;36m{set2}\033[0m: {set1 >= set2}")
# Subset and superset checks for overlaping sets
check_subset_superset(set_a, set_b)  # set_a is a subset of set_b, but not a superset
print("==="*8)
# Subset and superset checks for independent sets
check_subset_superset(set_a, set_c)  # set_a and set_c have no elements in common
```

Operations on iteratables:

- iteration with for loop
- unpacking with `*` operator
- `in` operator to check if an element is in the iteratable
- expand with `*` operator.

```python
# unpacking
datetime = (('04', "12", '2026'), (6, 30, "pm"))

(month, day, year), time = datetime

print(f"Unpacked datetime: month={month}, day={day}, year={year}, time={time}")
# or fully unpacking time as well
(month, day, year), (hour, minute, period) = datetime
print(f"Fully unpacked datetime: month={month}, day={day}, year={year}, hour={hour}, minute={minute}, period={period}")
```

`_`  is  a throw-away variable used when we want to ignore a value in unpacking or iteration. it has has the value of last assinged.

```python
a, _ = (1, 2)  # Using _ to ignore values during unpacking
print(f"After unpacking with throw-away variables: _ = {_}")
_, chr2 = "AB"
print(f"After unpacking characters with throw-away variable: _ = {_}")
```

```python
# * based unpacking
numbers = [1, 2, 3, 4, 5]
first, *middle, last = numbers
print(f"First: {first}, Middle: {middle}, Last: {last}")
print(f"Type of middle: {type(middle)}") # unpacking with * always gives a list even with only one element in middle.

*extra, last1, last2 = numbers
print(f"Extra: {extra}, Last1: {last1}, Last2: {last2}")

first, *extra = numbers
print(f"First: {first}, Extra: {extra}")

# or just to ignore the middle values
first, *_, last = numbers
print(f"First: {first}, Last: {last}")

# using the older datetime example for nested unpacking.

(month, *_), _ = datetime
print(f"Unpacked datetime with nested unpacking, throw away rest: month={month}, Last throw away _ = {_}")
```

```python
# expanding iterables with * operator
numbers = [1, 2, 3]
more_numbers = [4, 5]

combined = [*numbers, *more_numbers]
random_combination = [0, *numbers, 6, *more_numbers, 7]
print(f"Combined list using * operator: {combined}")
print(f"Random combination of numbers: {random_combination}")

# for dictionaries we use ** operator single  * will combine the keys and give us a set of keys
dict1 = {'a': 1, 'b': 2}
dict2 = {'c': 3, 'd': 4}

keys_combined = {*dict1, *dict2}
print(f"Combined keys using * operator: {keys_combined}") 
combined_dict = {**dict1, **dict2}
print(f"Combined dictionary using ** operator: {combined_dict}")
```

**Sequences:** A container in which we can add and remove items. Access via an integer index.

```python
from copy import deepcopy
a = [3, 4, 5]
print(f"Original list a: {a}")
b = [a]
print(f"List b referencing a: b = {b}")
a[0] = 123
print(f"After modifying a: a = {a}, b = {b}")
# reference extend as well
c = 2*b
print(f"Extending to c still references the same list b does: c = {c}")
a [0] = 1
print(f"After modifying a again: a = {a}, b = {b}, c = {c}")

# two ways ether use deep copy or use list

print("\033[1;34m===\033[0m"*8)
print("\033[1;36mCopying list a to b using list(a) and deep copy to c\033[0m")
a = [3, 4, 5]
print(f"Original list a: {a}")
b = [list(a)]
print(f"List b using list(a) to copy a: b = {b}")
c = deepcopy(b)
print(f"List c as a deep copy of b: c = {c}")
print(f"Before modifying a: a = {a}, b = {b}, c = {c}")
a[0] = 123
print(f"After modifying a: a = {a}, b = {b}, c = {c}")
```

### ANSI Escape Syntax for Terminal Formatting                                                                                                                                                
                                                                                                                                                                                                
  **Format:**                                                                                                                                                                                   
  \033[;m your text \033[0m                                                                                                                                                                     
                                                                                                                                                                                                
  - `\033` → ESC character (ASCII 27), starts the escape sequence                                                                                                                               
  - `[` → Control Sequence Introducer (CSI)                                                                                                                                                     
  - `<style>;<color>` → semicolon-separated SGR codes (order doesn't matter)                                                                                                                    
  - `m` → ends the formatting command                                                                                                                                                           
  - `\033[0m` → resets all formatting back to default                                                                                                                                           
                                                                                                                                                                                                
  **Common style codes:**                                                                                                                                                                       
  | Code | Effect      | Code | Effect        |
  |------|-------------|------|---------------|                                                                                                                                                 
  | 1    | Bold        | 4    | Underline     |                                                                                                                                                 
  | 2    | Dim         | 7    | Reverse       |
  | 3    | Italic      | 9    | Strikethrough |                                                                                                                                                 
                                                   

```python
# fun little tangent
# styling format for terminal output using ANSI escape codes
print("\033[1m Simple Bold text \033[0m")
print("\033[31m Red Colored text \033[0m")
print("\033[32m Green Colored text \033[0m")
print("\033[33m Yellow Colored text \033[0m")
print("\033[34m Blue Colored text \033[0m")
print("\033[35m Magenta Colored text \033[0m")
print("\033[36m Cyan Colored text \033[0m")
print("\033[1;3;4;30m Bold italic Black Colored text with underline\033[0m")
print("\033[1;35m Bold Magenta Colored text \033[0m")

```

Two way to slice:
- using `slice(i, j, stride)`
- or `sequence[i: j: stride]`

for both j in exlusive

```python
a = [*range(10)]
print(f"List to slice: {a}")
reverse_slice_exclusive = slice(6, 0, -1)
print(f"Reverse slice from index 6 to 0 exclusive with step -1: {a[reverse_slice_exclusive]}")
reverse_slice = slice(6, None, -1)
print(f"Reverse slice from index 6 to the end with step -1: {a[reverse_slice]}")
reverse_slice_full = slice(None, None, -2)
print(f"Reverse slice of the entire list with step -2: {a[reverse_slice_full]}")
```

Expansion operations on mutables

```python
example = [9, 8, 7, 6, 5, 4, 3, 2, 1, 0]
print(f"Original list: {example}")
# simple replacement of an element at index 1
example[1] = 12
print(f"After replacing element at index 1: {example}")
example[2:4] = [15, 16]
print(f"After replacing range from index 2 to 4 exclusive: {example}")
# range replacement from index 4 to 6 exclusive with step 2
example[4:] = [99, 88] # this will replace two elements of the range and discard rest.
print(f"After replacing range from index 4 exclusive with step 2: {example}")
example[5:6] = [77, 66, 55] # this will replace the range from index 4 to 6 exclusive and add one more element in between.
print(f"After replacing range from index 4 to 6 exclusive with step 2: {example}")
```

```python
# with stride we get value error if len of replacement does not match the slice
print(f"Original list for stride replacement: {example}")
print("Attempting to replace lenght 3 slice with lenght 2 list")
example = [11, 22, 33, 44, 55, 66, 77, 88, 99]
try:
    example[:6:2] = [1, 2]  # This will raise a ValueError because the slice has 3 elements but we are trying to assign only 2.
except ValueError as e:
    print(f"Error: {e}")
example[:6:2] = [1, 2, 3]  # This will work because the slice has 3 elements and we are assigning 3 values.
print(f"After matching right side length 2: {example}")
# or 
example[:4:2] = [1, 2]
print(f"OR matching left side length: {example}")
```

Geneartor Introduction (Later a whole chapter):
- generators are like lists comprehensions, but they compute on demand
- are exhausted once evaluated
- can be defined using `()` instread of `[]` in a comprehension
- function with yield are generators
- it only looks like list comprehension most of the list operations donot work.


```python
nums = [1, 2, 3, 4, 5]

squared = [x**2 for x in nums]

# we can create a generator expression by using parentheses instead of square brackets, this will not compute.
squared_gen = (x**2 for x in nums)
print(f"Squared list: {squared}")
print(f"Squared generator: {squared_gen}")
# in order to get the values
for value in squared_gen:
    print(f"Squared generator value: {value}")
# Generators once evaluated are exhausted, this will raise error
try:
    next_value = next(squared_gen)
except StopIteration:
    print("Generator is exhausted.")

# we can get items with next
squared_gen = (x**2 for x in nums)
print(f"Next value from squared generator: {next(squared_gen)}")
print(f"Next value from squared generator: {next(squared_gen)} \n...")
```
