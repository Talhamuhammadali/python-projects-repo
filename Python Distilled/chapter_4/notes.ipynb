```python
import inspect

a = 10
inspection = inspect.getmembers(a)
inspection = [item for item in inspection if not item[0].startswith('__')]
formated = [f"{item[0]}: {item[1]}" for item in inspection]
header = "name | value"
final_output = "\n".join([header] + formated)
print(final_output)
```

Bit of tangent Exploring inspect module. It has following methods:

- `getmembers`: Returns all the members of an object in a list of tuples.
- `get_annotations`: Returns a dictionary of annotations for an object.


Note: When inspecting any annotation it should be defined in the module, class or function for it to be visible in inspection.

```python
import collections
import dataclasses
from langchain_core.messages import AIMessage

# Small test
print(inspect.collections is collections)

# ---
# Testing annotations
num: int = 10
print(collections.__annotations__) # Not defined for collections module so should be empty
print(inspect.get_annotations(collections)) # same as above
try:
    print(inspect.get_annotations(num))  # Should give error as objs do not have annotations
except Exception as e:
    print(f"Error: {e}")
    # So objects donot have annotations only modules

print(inspect.get_annotations(dataclasses))  # empty 

print(inspect.get_annotations(AIMessage))  
```

```python
# Custom example
import annotated_module
from annotated_module import add_numbers, Point

print(inspect.get_annotations(Point))  # Should give the annotations defined in the class
print(inspect.get_annotations(add_numbers))  # Should give the annotations defined in the function
print(inspect.get_annotations(annotated_module))  # Should give the annotations defined in the module

point = Point(1, 2)
try:
    print(inspect.get_annotations(point))  # Should give error as objs do not have annotations
except Exception as e:
    print(f"Error: {e}")
# but directly inspecting the class should work
print(point.__annotations__) # this is because now this is now an instance of the class not the class itself. this is
# inherited from Point class.
```

Something cool i learnt:

- `get_annotations()` method can only inspect a module, a callable(functions, methoods), or a class. It cannot inspect an instance of a class.

- for classes it stays on attributes level
- for functions/methods it stays on parameters and return type level
- for modules it stays on module level typed variables.

See the example above. In short annotation is just a record of type hints.

*Why get_annotations fail on instances?*

Python is not strictly typed language. Therefore, even if the source types of instance attributes are defined, python does not enforce them. So even if `Point` says x an y are floats we can still assign them to strings. Therefore,
annotations are fetched from the class as that is the main contract/source of truth.
