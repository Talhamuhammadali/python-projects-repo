```python
import pandas as pd
pd.set_option('display.max_rows', None)
pd.set_option('display.max_columns', None)
with open("market_history.csv", "r") as file:
    df = pd.read_csv(file)
columns = df.columns.to_list()
print(len(df))
cent_price_col = columns[5]
game_mask = df[columns[1]] == "Counter-Strike 2"
item_name = columns[0]
action_type = columns[6]
items_with_price = df[game_mask][[item_name, action_type, cent_price_col]].sort_values(
    by=cent_price_col, ascending=False
)
items_with_price["Price USD"] = items_with_price[" Price in Cents"].divide(100)
agg_sales = items_with_price.groupby(action_type)[[cent_price_col, "Price USD"]].sum()
agg_sales.to_csv("sales_total.csv")
# items_sales_purchase = items_with_price.groupby([item_name, action_type])[["Price USD"]].sum().sort_values(by=item_name, ascending= False)
# items_sales_purchase.to_csv("sales_purchase_by_items.csv")
# print(items_sales_purchase)
items_sales_purchase = items_with_price.groupby([item_name, action_type]).agg({
    "Price USD": ["sum", "count"]
}).sort_values(by=item_name, ascending=False)

items_sales_purchase.to_csv("sales_purchase_by_items.csv")
print(items_sales_purchase)
```
