# Calculating Compounding Effect 

```python
principal = 740000
monthly_sip = 71000
monthly_profit = 0.18/12
num_yrs = 20
year = 1
print("year", "new_princ")
while year <= num_yrs:
    for _ in range(12):
        principal = (principal + monthly_sip) * (1 + monthly_profit)
    print(year, f"{principal/10**6:.3f}m")
    year += 1
```

```python
target_corpus = 30000000  # 3 crores
annual_return = 0.20
monthly_return = annual_return / 12
years = 15

# SIP formula: PMT = FV * r / ((1+r)^n - 1)
months = years * 12
monthly_sip = target_corpus * monthly_return / ((1 + monthly_return)**months - 1)
print(f"Monthly SIP needed: ₹{monthly_sip:,.0f}")
```

```python
from pydantic import BaseModel, ValidationError, EmailStr, Field
from streamlit import user

class User(BaseModel):
    id: int
    name: str
    title: str = Field(..., min_length=3, max_length=50)
    age: int = Field(..., gt=0, lt=50)
    email: EmailStr
    
Examples = [
    {
        "id": 123,
        "name": "John Doe",
        "title": "Software Engineer",
        "age": 25,
        "email": "john.doe@example.com"
    },
    {
        "id": 123,
        "name": "John Doe", 
        "title": "Manager",
        "age": 35,  
        "email": "john.doeexample.com"  # Invalid email
    },
    {
        "id": 124,
        "name": "Jane Doe",
        "title": "Director",
        "age": 90, # Invalid age
        "email": "jane.doe@example.com"
    },
    {
        "id": 125,
        "name": "Alice Smith",
        "title": "HR in you ass bitch ass dumb ass mf"*2,  # Invalid title length
        "age": 30,
        "email": "alice.smith@example.com"
    }
]
for example in Examples:
    try:
        user = User(**example)
        print(user)
    except ValidationError as e:
        print("===Validation error===")
        err_formatted = "\n".join([f"{err['loc'][0]}: {err['msg']}" for err in e.errors()])
        print(err_formatted)
```
