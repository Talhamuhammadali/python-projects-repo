# Calculating Compounding Effect 

```python
principal = 740000
monthly_sip = 71000
monthly_profit = 0.18/12
num_yrs = 20
year = 1
print("year", "new_princ")
while year <= num_yrs:
    for _ in range(12):
        principal = (principal + monthly_sip) * (1 + monthly_profit)
    print(year, f"{principal/10**6:.3f}m")
    year += 1
```

```python
target_corpus = 30000000  # 3 crores
annual_return = 0.20
monthly_return = annual_return / 12
years = 15

# SIP formula: PMT = FV * r / ((1+r)^n - 1)
months = years * 12
monthly_sip = target_corpus * monthly_return / ((1 + monthly_return)**months - 1)
print(f"Monthly SIP needed: ₹{monthly_sip:,.0f}")
```
