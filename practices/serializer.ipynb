## Cleaing an object to be serializable
Example below is of an object that cannot be serialized

```python
import json 
from decimal import Decimal
from typing import Any

model_dict = [
  {
    "model_name": "gpt-5-mini",
    "temperature": Decimal(1)
  },
  {
    "model": "claude-sonnet-4-5-20250929",
    "temperature": 1,
    "max_tokens": Decimal(64000),
    "thinking": {
      "budget_tokens": Decimal(10000),
      "type": "enabled"
    }
  }
]
# json.dumps(model_dict)
model_dict
```

```python
def make_serializeable(obj: Any) -> Any:
    if isinstance(obj, list):
        return [make_serializeable(item) for item in obj]
    if isinstance(obj, dict):
        return({k:make_serializeable(v) for k,v in obj.items()})
    if isinstance(obj, Decimal):
        return int(obj)
    return obj

clean_dict = make_serializeable(model_dict)
model_dict, clean_dict
```

```python
json.dumps(clean_dict)
```
