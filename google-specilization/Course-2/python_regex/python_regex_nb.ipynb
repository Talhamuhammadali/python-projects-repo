**Wildcards and Charater Classes**

- `[]` is to specify range of Character classes to be used as wildcard. This is also called a group.
- `^` in side the group stands for ignore/dont match.
- `|` match multiple patterns on **or** condition.
- `re.findall()` to give a list of all the matches.

```python
print(re.search(r"[Pp]ython", "Python")) # matches both P and p at the start of python
print(re.search(r"[a-z]way", "The end of the highway")) # matches alphabets [a-z] at the on the first character and ending with way
print(re.search(r"[a-z]way", "What a way to go")) # matches alphabets [a-z] at the on the first character and ending with way
print(re.search("cloud[a-zA-Z0-9]", "cloudy")) # matches alphabets both cases [a-z] and numbers 0-9 as the last character at the end of cloud
print(re.search("cloud[a-zA-Z0-9]", "cloud9")) # matches alphabets both cases [a-z] and numbers 0-9 as the last character at the end of cloud
print(re.search(r"[^a-zA-Z]", "This is a sentence with spaces.")) # donot matches alphabets both cases [a-z]
print(re.search(r"[^a-zA-Z ]", "This is a sentence with spaces.")) # donot matches alphabets both cases [a-z]

print(re.search(r"cat|dog", "I like cats.")) # cat or dog match
print(re.search(r"cat|dog", "I love dogs!"))
print(re.search(r"cat|dog", "I like both dogs and cats."))

print(re.search(r"cat|dog", "I like cats."))
print(re.search(r"cat|dog", "I love dogs!"))
print(re.search(r"cat|dog", "I like both dogs and cats."))
print(re.findall(r"cat|dog", "I like both dogs and cats."))
```

**Repetition Qualifiers**

- `.*` Repeat any character match as many there are that match the pattern.
- `[]*` Repeat characters only sepcified in the group.
- `+` one or more occurrences of the preceding element.
  
  **General behavior of +:**
    - `a+` matches: "a", "aa", "aaa", etc. (but not "")
    - `[abc]+` matches any sequence of one or more a's, b's, or c's in any order
    - `\d+` matches one or more digits
  
- `?` zero or one occurrence of the preceding element.

```python
# *
print(re.search(r"Py.*n", "Pygmalion")) # match pattern where Py followed by any characters(.*) and finally followed by n
print(re.search(r"Py.*n", "Python Programming")) # same as before
print(re.search(r"Py[a-z]*n", "Python Programming")) # match only lower case a-z
print(re.search(r"Py[a-z]*n", "Pyn")) # match only lower case a-z
# +
#  r"o+l+" matches one or more 'o's followed by one or more 'l's
print(re.search(r"o+l+", "goldfish"))
print(re.search(r"o+l+", "woolly"))
print(re.search(r"o+l+", "boil"))
# ?
print(re.search(r"p?each", "To each their own"))
print(re.search(r"p?each", "I like peaches"))
```

```python
# ? examples
print("\nExample 1: colou?r (matches 'color' or 'colour')")
regex1 = r"colou?r"
print("'color':", bool(re.search(regex1, "color")))      # True
print("'colour':", bool(re.search(regex1, "colour")))    # True
print("'colouur':", bool(re.search(regex1, "colouur")))  # False (too many u's)

print("\nExample 2: Feb(ruary)? (matches 'Feb' or 'February')")
regex2 = r"Feb(ruary)?"
print("'Feb':", bool(re.search(regex2, "Feb")))            # True
print("'February':", bool(re.search(regex2, "February")))  # True
print("'Febr':", bool(re.search(regex2, "Febr")))          # True (because it finds 'Feb')

# Let's check exact matches for Example 2
print("\nExample 2b: Exact matches with ^...$ anchors")
regex2b = r"^Feb(ruary)?$"
print("'Feb' (exact):", bool(re.search(regex2b, "Feb")))            # True
print("'February' (exact):", bool(re.search(regex2b, "February")))  # True
print("'Febr' (exact):", bool(re.search(regex2b, "Febr")))          # False (now it fails)

print("\nExample 3: \\d+ and \\d? (number patterns)")
regex3 = r"\d+\.\d?"
print("'5.':", re.search(regex3, "5."))       # Matches
print("'5.2':", re.search(regex3, "5.2"))     # Matches
print("'5.23':", re.search(regex3, "5.23"))   # Matches "5.2"
# Let's see what exactly was matched in the last case
match = re.search(regex3, "5.23")
if match:
    print("In '5.23', matched portion:", match.group(0))

print("\nExample 4: Words ending with optional 's'")
regex4 = r"\bbook(s)?\b"
print("'I read a book':", bool(re.search(regex4, "I read a book")))     # True
print("'I read books':", bool(re.search(regex4, "I read books")))       # True
print("'I read booking':", bool(re.search(regex4, "I read booking")))   # False

print("\nExample 5: Optional punctuation")
regex5 = r"Hello,? world!?"
print("'Hello world':", bool(re.search(regex5, "Hello world")))         # True
print("'Hello, world':", bool(re.search(regex5, "Hello, world")))       # True
print("'Hello world!':", bool(re.search(regex5, "Hello world!")))       # True
print("'Hello, world!':", bool(re.search(regex5, "Hello, world!")))     # True
```

**Escaping Charaters**

- `\` if you want to match special character that have defined function in regex. i.e `.,  [], ?, +, *` are example of special characters
- `\w` matches any alphanumeric charaters i.e numbers, alphabates and underscores.
- `\s` for space, tab and newline.
- `\d` for digits.
- `\b` for word boundaries.

```python
# \
print(re.search(r".com", "welcome")) # matches any character
print(re.search(r"\.com", "welcome")) # matches .com
print(re.search(r"\.com", "mydomain.com")) # example with match
# \w
print(re.search(r"\w*", "This is an example"))
print(re.search(r"\w*", "And_this_is_another"))
```

**Advance Regex**

*Capturing Groups*

`()` groups multiple characters together and treats them as a single unit

```python
result = re.search(r"^(\w*), (\w*)$", "Lovelace, Ada")
print(result)
print(result.groups())
print(result[0])
print(result[1])
print(result[2])
"{} {}".format(result[2], result[1])
```

```python
def rearrange_name(name):
    result = re.search(r"^([\w \.-]*), ([\w \.-]*)$", name)
    if result is None:
        return name
    return "{} {}".format(result[2], result[1])
print(rearrange_name("Lovelace, Ada"))
print(rearrange_name("Ritchie, Dennis"))
print(rearrange_name("Hopper, Grace M."))
```

*Repetition qualifiers*

`{n, }` Match the pattern to n or more characters per group.

```python
print(print(re.search(r"[a-zA-Z]{5}", "a ghost")))
print(re.findall(r"[a-zA-Z]{5}", "a scary ghost appeared")) # match in-case sensitive alphabets 5 times into groups
print(re.findall(r"\b[a-zA-Z]{5}\b", "A scary ghost appeared boo")) # same but its a 5 charater word. by speciing word boundary
print(re.findall(r"\w{5,}", "I really like strawberries omgomgomgomgomgomgomg"))  # match words with 5 or more charaters per group
print(re.findall(r"\w{5,10}", "I really like strawberries omgomgomgomgomgomgomg")) # match words with 5 - 10 characters 
print(re.findall(r"s\w{,20}", "I really like strawberries so much")) # group with up to 20 charaters starting with s
```

```python
import re
def long_words(text):
  pattern = r"[a-zA-Z]{7,}"
  result = re.findall(pattern, text)
  return result

print(long_words("I like to drink coffee in the morning.")) # ['morning']
print(long_words("I also have a taste for hot chocolate in the afternoon.")) # ['chocolate', 'afternoon']
print(long_words("I never drink tea late at night.")) # []
```

```python
import re
def extract_pid(log_line):
    regex = r"\[(\d+)\].*\b([A-Z]+)\b"
    result = re.search(regex, log_line)
    if result is None:
        return None
    return "{} ({})".format(result[1], result[2])

print(extract_pid("July 31 07:51:48 mycomputer bad_process[12345]: ERROR Performing package upgrade")) # 12345 (ERROR)
print(extract_pid("99 elephants in a [cage]")) # None
print(extract_pid("A string that also has numbers [34567] but no uppercase message")) # None
print(extract_pid("July 31 08:08:08 mycomputer new_process[67890]: RUNNING Performing backup")) # 67890 (RUNNING)
```

The phone number field contains U.S. phone numbers and needs to be modified to the international format, with +1- in front of the phone number. The rest of the phone number should not change. Fill in the regular expression, using groups, to use the transform_record() function to do that.

```python
import re
def transform_record(record):
  new_record = re.sub(r",(\d+)-", r",+1-\1-",record)
  return new_record

print(transform_record("Sabrina Green,802-867-5309,System Administrator")) 
# Sabrina Green,+1-802-867-5309,System Administrator

print(transform_record("Eli Jones,684-3481127,IT specialist")) 
# Eli Jones,+1-684-3481127,IT specialist

print(transform_record("Melody Daniels,846-687-7436,Programmer")) 
# Melody Daniels,+1-846-687-7436,Programmer

print(transform_record("Charlie Rivera,698-746-3357,Web Developer")) 
# Charlie Rivera,+1-698-746-3357,Web Developer
```

The multi_vowel_words() function returns all words with 3 or more consecutive vowels (a, e, i, o, u). Fill in the regular expression to do that.

```python
import re
def multi_vowel_words(text):
  pattern = r"\b\w*[aeiou]{3,}\w*\b"
  result = re.findall(pattern, text)
  return result

print(multi_vowel_words("Life is beautiful")) 
# ['beautiful']

print(multi_vowel_words("Obviously, the queen is courageous and gracious.")) 
# ['Obviously', 'queen', 'courageous', 'gracious']

print(multi_vowel_words("The rambunctious children had to sit quietly and await their delicious dinner.")) 
# ['rambunctious', 'quietly', 'delicious']

print(multi_vowel_words("The order of a data queue is First In First Out (FIFO)")) 
# ['queue']

print(multi_vowel_words("Hello world!")) 
# []
```

```python
def parse_city_country(text):
  pattern = r", |\. " #enter the regex pattern here
  result = re.split(pattern, text) #enter the re method  here
  if len(result) != 2:
    return ""
  return result[0]#return the correct capturing group

print(parse_city_country("Paris, France")) # should return Paris
print(parse_city_country("Mumbai, India")) # should return Mumbai
print(parse_city_country("Rio de Janeiro. Brazil")) # should return Rio de Janeiro
print(parse_city_country("Tokyo! Japan"))  # result should be blank
```
