# Testing in Python
**What is Testing?**

When a program grows in complexity, it becomes difficult to test if all the functionalities are working as intended. So, we have software testing; process of evaluating computer code to determind whether it works as intended or not.

*Important testing concepts*
- Unit Tests
- Automated Tests
- integration Tests
- Test-Driven Development

**Test cases**

Test cases are scenarios where a system has possiblity of not performing as its expected too. A test case always expects the same output for each case if program doesnt produce such output it means code is buggy. Test cases helps us evalute the code quickly for possible edge cases. More cases are added if issue are discovered. This is called automated testing.

## Unit Test

Unit tests are run on components or by isolating units of code to ensure correctness. This creates reliable code as error are discovered early.

**Concepts**

- **Test fixture:** This refers to preparing to perform one or more tests. In addition, test fixtures also include any actions involved in testing cleanup. This could involve creating temporary or proxy databases, directories, or starting a server process.

- **Test case:** This is the individual unit of testing that looks for a specific response to a set of inputs. If needed, TestCase is a base class provided by unittest and can be used to create new test cases.

- **Test suite:** This is a collection of test cases, test suites, or a combination of both. It is used to compile tests that should be executed together.

- **Test runner:** This runs the test and provides developers with the outcome’s data. The test runner can use different interfaces, like graphical or textual, to provide the developer with the test results. It can also provide a special value to developers to communicate the test results.

**`unittest` Module**

This python module provides functionality to create unit test scripts in python


### Use case
Create a FlowerShop class that helps manage a small flower business. The class should handle flower arrangements with the following requirements:

1. The class should be initialized with a bouquet_type (options: "seasonal", "classic", "luxury") and size (options: "small", "medium", "large").
2. The price should be calculated based on:

    - Base price: 15 for "luxury", 12 for "classic", and 10 for "seasonal"
    - Size adjustment: +0 for "small", +3 for "medium", +6 for "large"


3. The shop should allow adding optional flowers to the arrangement, with each additional flower increasing the price by 2.
4. There should be a method to check what supplies are needed for the arrangement.
5. There should be a method to check the final price.

```python

import unittest
from typing import Optional, Literal, List
from enum import Enum

class Size(str, Enum):
    SMALL = "small"
    MEDIUM = "medium"
    LARGE = "large"
    
class BouquetTypes(str, Enum):
    CLASSIC = "classic"
    SEASONAL = "seasonal"
    LUXURY = "luxury"

class FlowerShop:
    def __init__(self, bouquet_type: str, size: str):
        
        self.bouquet_type = bouquet_type
        self.size = size
        self.additional_flowers = []
        self.flower_catalog = {
            BouquetTypes.CLASSIC: [
                "Rose",
                "Lily",
                "Carnation"
            ],
            BouquetTypes.SEASONAL: [
                "Tulip",
                "Daffodil",
                "Sunflower"
            ],
            BouquetTypes.LUXURY: [
                "Orchid",
                "Peony",
                "Hydrangea"
            ]
        }
        
        # price calculation
        self.price = 15 if bouquet_type == BouquetTypes.LUXURY else 12 if bouquet_type == BouquetTypes.CLASSIC else 10
        self.price += 6 if size == Size.LARGE else 3 if size == Size.MEDIUM else 1
        
    def add_flowers(self, flower_name: str):
        available, type = self.check_supply(flower_name=flower_name)
        if available:
            self.additional_flowers.append(flower_name)
            additional_cost = 2 if type == BouquetTypes.CLASSIC else 5
            self.price += additional_cost
            print(f"{type.name} flower added cost {additional_cost}")
        else:
            print(type)
        
    def check_supply(self, flower_name: str) -> tuple:
        for type, flowers in self.flower_catalog.items():
            if flower_name in flowers:
                return(True, type)
        return (False, "Not available")
    
    def check_price(self):
        return self.price
    

bouquet = FlowerShop(bouquet_type="luxury", size="large")
print(bouquet.check_price())
bouquet.add_flowers("Roses")
bouquet.check_price()
```

```python
# Test cases using unittest
import unittest

class TestFlowerShop(unittest.TestCase):
    def test_create_bouquet(self):
        bouquet = FlowerShop(bouquet_type="classic", size="large")
        self.assertEqual(bouquet.size, "large")
        self.assertEqual(bouquet.bouquet_type, "classic")
        self.assertEqual(bouquet.price, 18)

unittest.TextTestRunner().run(unittest.TestLoader().loadTestsFromTestCase(TestFlowerShop))
```

**pytest**

Pytests are written with functions that use the operation, assert().

```python
def divide(a, b):
	assert b != 0, "Cannot divide by zero"
	return a / b
try:
	divide(12, 0)
except Exception as e:
	print(e)
```

```python
import pytest
class Fruit:
    def __init__(self, name):
        self.name = name
        self.cubed = False


    def cube(self):
        self.cubed = True


class FruitSalad:
    def __init__(self, *fruit_bowl):
        self.fruit = fruit_bowl
        self._cube_fruit()


    def _cube_fruit(self):
        for fruit in self.fruit:
            fruit.cube()


# Arrange
@pytest.fixture
def fruit_bowl():
    return [Fruit("apple"), Fruit("banana")]


def test_fruit_salad(fruit_bowl):
    # Act
    fruit_salad = FruitSalad(*fruit_bowl)


    # Assert
    assert all(fruit.cubed for fruit in fruit_salad.fruit)
```

```python
import unittest

class TestRearrange(unittest.TestCase):
    #  any method starting with test_ will be considered a test case
    def test_basic(self):
        testcase = "Lovelace, Ada"
        expected = "Ada Lovelace"
        self.assertEqual(rearrange_name(testcase), expected)
        
    def test_empty(self):
        testcase = ""
        expected = ""
        self.assertEqual(rearrange_name(testcase), expected)
    
    def test_double_name(self):
        testcase = "Hopper, Grace M."
        expected = "Grace M. Hopper"
        self.assertEqual(rearrange_name(testcase), expected)

    def test_one_name(self):
        testcase = "Voltaire"
        expected = "Voltaire"
        self.assertEqual(rearrange_name(testcase), expected)

# Run the tests
test_suite = unittest.TestLoader().loadTestsFromTestCase(TestRearrange)
unittest.TextTestRunner().run(test_suite)
```
