# Python Error Handling: Comprehensive Practice

## Why error handle 
- Continue running despite encountering errors
- Provide meaningful feedback instead of cryptic error messages
- Perform cleanup operations (like closing files) even when errors occur
- Implement fallback behaviors when operations fail

````python
                                                    # Basic Exceptions Example
try:
    user_input = input("Enter Number:")
    num = float(user_input)
    result = 100/num
    print(f"100 divided by {num} gives: {result}")
except ValueError:
    print("Error: You must enter a valid number.")
except ZeroDivisionError:
    print("Error: You cannot divide by zero.")

# Possible Exceptions here are:
# ZeroDivisionError: if we divide by zero
# ValueError: if user enters non-numberic item
# KeyboardInterrupt: Key board intrupt ctrl + c
````

## Exceptions in Python
Python has a rich hierarchy of built-in exception types. All exceptions inherit from the base Exception class, which itself inherits from BaseException. Common exceptions include:

- `SyntaxError`: Invalid Python syntax
- `TypeError`: Operation applied to inappropriate type
- `ValueError`: Operation received correct type but inappropriate value
- `NameError`: Local or global name not found
- `IndexError`: Sequence index out of range
- `KeyError`: Dictionary key not found
- `FileNotFoundError`: Requested file doesn't exist
- `ZeroDivisionError`: Division or modulo by zero
- `ImportError`: Import statement failed
- `IOError`: Input/output operation failed

````python
# Uncomment lines to see error
# IndexError
try:
    alist = [1, 2, 3]
    print(alist[4])
except Exception as ex:
    print(ex)

# KeyError
try:
    adict = {'a':1, "b": 2}
    print(adict["c"])
except Exception as ex:
    print(ex)

# TypeError
try:
    text = "hello"
    print(text + 5)
except Exception as ex:
    print(ex)

# ValueError
try:
    num = int("10.0") # cannot convert floats
    num = int("abc") # Cannt convert non numeric string
except Exception as ex:
    print(ex)

# ZeroDivisionError
try:
    result =  10/ 0
except Exception as ex:
    print(ex)

# More Example
# a = 10 / (5 - 5) # ZeroDivisionError
# b = "hello" / 2 # TypeError
# c = [1, 2, 3].remove(4) # ValueError
# d = int(input("Enter a number: "))  # User enters "xyz" ValueError
# e = open("nonexistent_file.txt", "r") # FileNotFoundError error
````

## Basic Error Handling with try/except
```python
try:
    # Main logic
    pass
except: # with exception type
    pass # fail safe logic
```
When the try block executes:

- If no exception occurs, the except block is skipped
- If an exception occurs, Python looks for a matching except block
- If a matching except block is found, it executes and then continues after the try/except structure
- If no matching except block is found, the exception propagates upward

````python
try:
    user_input = input("Enter number greater then zero:")
    num = int(user_input)
    result = 100/num
    print(f"100 divided by {num} is {result}")

except ZeroDivisionError as z_err:
    print(f"Error: {z_err}")
except ValueError as v_err:
    print(f"Error: {v_err}")
````

````python
# If not specfice exception that could occure is detemined its recommended to use basic exception clas
try:
    # Risky code
    user_input = input("Enter a number: ")
    number = int(user_input)
    result = 100 / number
    print(f"100 divided by {number} is {result}")
except Exception as e:
    # Gets the exception object
    print(f"An error occurred: {e}")
````

````python
# Practice implementing try catch when reading file
def safe_open(filename: str):
    try:
        with open(filename, 'r') as file:
            return file.readlines()
    except FileNotFoundError as err:
        print(err)
safe_open("README.md")
safe_open("test")
````

````python
def str_to_int_conversion(value: str) -> int:
    try:
        integer = int(value)
        return integer
    except ValueError as v_err:
        return v_err
print(str_to_int_conversion(0))    
print(str_to_int_conversion("02"))    
print(str_to_int_conversion("23"))
print(str_to_int_conversion("2.3"))        
````
